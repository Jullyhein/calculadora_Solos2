```python
from openpyxl import load_workbook

# Caminho para o arquivo Excel
file_path = os.path.join(DOWNLOAD_FOLDER, file_name)  # Substitua pelo caminho correto do arquivo

# Carregar a planilha com os valores calculados
wb = load_workbook(file_path, data_only=True)  # 'data_only=True' pega valores calculados, não fórmulas
sheet = wb.active  # Seleciona a primeira aba ativa

# Iterar sobre a coluna ou célula com fórmulas
valores = []
for row in sheet.iter_rows(min_row=2, max_row=sheet.max_row, min_col=19, max_col=19):  # Ajuste os índices da coluna
    for cell in row:
        valores.append(cell.value)  # Adiciona os valores calculados à lista

# Exibe os valores calculados
print("Valores calculados:")
for valor in valores:
    if valor is not None:  # Verifica se o valor não é None
        print(f"{valor:.2f}")
    else:
        print("None")

# Soma os valores, se necessário
total = sum(filter(None, valores))  # Filtra valores None antes de somar
print(f"Total da coluna: {total:.2f}")
```

```python
df['PREÇO DA MESA /MÓDULO'].fillna(0, inplace=True)
print("Coluna 'PREÇO DA MESA /MÓDULO' após tratamento:")
print(df['PREÇO DA MESA /MÓDULO'])
```

```python
# Diagnóstico inicial dos valores originais
print("Valores originais na coluna 'PREÇO DA MESA /MÓDULO':")
print(df['PREÇO DA MESA /MÓDULO'].head(20))

# Limpeza avançada da coluna
if 'PREÇO DA MESA /MÓDULO' in df.columns:
    df['PREÇO DA MESA /MÓDULO'] = (
        df['PREÇO DA MESA /MÓDULO']
        .astype(str)
        .str.strip()  # Remove espaços extras
        .str.replace(r"[^\d.,-]", "", regex=True)  # Remove caracteres não numéricos
        .str.replace(",", ".")  # Substitui vírgulas por pontos
    )

    # Converter para float
    df['PREÇO DA MESA /MÓDULO'] = pd.to_numeric(df['PREÇO DA MESA /MÓDULO'], errors='coerce')

    # Exibir valores únicos após a limpeza
    print("Valores após limpeza e conversão:")
    print(df['PREÇO DA MESA /MÓDULO'].unique())

    # Verificar linhas com NaN
    nan_rows = df[df['PREÇO DA MESA /MÓDULO'].isna()]
    print("Linhas com valores NaN após limpeza:")
    print(nan_rows)
else:
    print("Coluna 'PREÇO DA MESA /MÓDULO' não encontrada.")
```
